# 03 - Reproduce all figures from the archived data

Runs `scripts/make_figures.py`, which calls the figure scripts in `experiments/` on the archived data in `results/` (nothing is
recomputed; about three minutes, most of it for Fig. 6) and writes `paper/figures/`. The figures are then displayed in paper order.
The table in `docs/FIGURES.md` maps each figure to its script, its data, and the scripts that generated the data.

```python
import sys, json
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path[:0] = [str(ROOT / 'code'), str(ROOT / 'experiments')]
np.set_printoptions(precision=4, suppress=True)
import subprocess
from IPython.display import Image, display, Markdown
```

```python
print(subprocess.run([sys.executable, str(ROOT / 'scripts' / 'make_figures.py')], cwd=ROOT, capture_output=True, text=True).stdout)
```

```python
for f in ['Fig01_schematic', 'Fig02_anatomy', 'Fig03_partial', 'Fig04_temperature', 'Fig05_sizes', 'Fig06_law', 'Fig07_families', 'Fig08_protocol', 'Fig09_controls', 'Fig10_larger', 'Fig11_learned', 'FigA1_E0_model', 'FigA2_E0', 'FigA3_tools', 'FigA4_static_law']:
    display(Markdown(f'**{f}**')); display(Image(filename=str(ROOT / 'paper' / 'figures' / (f + '.png')), width=900))
```
